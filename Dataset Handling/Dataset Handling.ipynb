3. Opening a CSV File From an URL

```python
import requests
import pandas as pd
from io import StringIO

url = ""
headers = {"User-Agent": "Mozilla/5.0"}

req = requests.get(url, headers=headers)
data = StringIO(req.text)

df = pd.read_csv(data)
print(df)
```
